```python
#test environment variable
import os
from dotenv import load_dotenv
```

```python
#load variable from local .env
load_dotenv()
Api_key = os.getenv("GROQ_API_KEY")
Model = os.getenv("GROQ_MODEL")

print("Api Key loaded successfully: ", Api_key is not None)
print("Active Model: ", Model)
```

```python
#Make Api request
import requests

url = "https://api.groq.com/openai/v1/chat/completions"
headers = {
    "Authorization": "Bearer " + Api_key,
    "Content-Type": "application/json"
}

userprompt= input("What would you like to ask the AI")

body = {"model": Model,
        "messages": [
            {"role": "user", "content": """Act as an Agri-Tech startup consultant.
Generate 5 innovative, practical business ideas for smallholder
farmers in Ogun State, Nigeria. 

For each idea include:
1. Name & Tagline
2. Problem Solved
3. Tech Solution (must account for low smartphone/internet access, e.g., USSD/SMS/Voice)
4. Revenue Model
5. How to reach local Ogun farmers

Additional user request:{userprompt}"""
}
]
        
        }
```

```python
#Send the Post request
result = requests.post(url, headers=headers, json=body)
print("key length: ", len(Api_key) if Api_key else "Not Found")

#check the http status code
print("HTTP status code: ", result.status_code)

```

```python
#convert the raw response into a python dictionary
data = result.json()

#navigate the nested json structure
answer = data["choices"][0]["message"]["content"]

print("---Welcome to the Business Idea Generator---")
print(answer)


#extract token usage statistics
tokens = data["usage"]["total_tokens"]

print

print(f"\ntotal Token Used: {tokens}")

```
