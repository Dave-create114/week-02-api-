```python
#test environment variable
import os
from dotenv import load_dotenv
```

```python
#load variable from local .env
load_dotenv()
Api_key = os.getenv("GROQ_API_KEY")
Model = os.getenv("GROQ_MODEL")

print("Api Key loaded successfully: ", Api_key is not None)
print("Active Model: ", Model)
```

```python
#Make Api request
import requests
import json

url = "https://api.groq.com/openai/v1/chat/completions"
headers = {
    "Authorization": "Bearer " + Api_key,
    "Content-Type": "application/json"
}
prompts = {
    "V1": "Write a product description for a solar-powered backpack.",
    
    "V2": "You are an expert E-Commerce Copywriter. Write a product description "
           "for a solar-powered backpack targeting tech-savvy outdoor commuters in Lagos.",
    
    "V3": """You are an expert Copywriter. Write a product description for a solar-powered "
           backpack targeting commuters.
Tone: Professional & inspiring
Length: Under 80 words
Format: You must respond strictly with a valid JSON object. Do not include any markdown formatting, wrappers, or text outside the JSON object."
Example Output:{"product_name:", "tagline:","bullet_points:[".."."..",]}""
JSON Schema:
           {"product_name":"string",
                    "tagline": "string",
                      "bullet_points": ["string"],
                      "target_audience": "string"
}"""
                   

}
def groq_assist(prompts):
    body= {"model": Model,
        "messages": [
            {"role": "user", "content":prompts}
        ],
        "temperature":0.7
    }
    response = requests.post(url, headers=headers, json=body) 
    response.raise_for_status()
    return response.json()["choices"][0]["message"]["content"]
    
  
```

```python
def score_response(version):
    scores = {
       "V1": {"clarity": 5, "specificity": 3, " structure":3},
       "V2": {"clarity": 8, "specificity": 8, " structure":6},
       "V3": {"clarity": 9, "specificity": 9, " structure":10}
    }
    return scores [version]


    
   
```

```python
results = {}

for version in ("V1", "V2", "V3"):
    response = groq_assist(prompts[version])
    scores = score_response(version)
    average = round(sum(scores.values()) / len(scores), 1)

    print(f"\n--- Version {version} ---")
    print("Response:")
    print(response)
    print("Scores:")
    for category, score in scores.items():
        print(f"  {category.strip()}: {score}/10")
    print(f"  Average: {average}/10")
```
